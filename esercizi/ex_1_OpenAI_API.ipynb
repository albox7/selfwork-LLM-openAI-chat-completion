**Traccia:** Scrivi un prompt che traduca un testo dall'italiano all'inglese e viceversa. Usa il prompt per tradurre una breve descrizione di un prodotto.

**Obiettivo:** Sviluppare un prompt che sia adatto alla traduzione fluida tra italiano e inglese.

```python

# TODO Inizializza il client di Open AI
import os, openai
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env') # Read the .env file

# Configure OpenAI client
openai.api_key = os.environ['OPENAI_API_KEY']

client = OpenAI()
```

```python
input_text = input("Enter text: ")

complention = client.chat.completions.create(
	model="gpt-3.5-turbo",
	messages=[
		{
			"role": "system",
			"content": "Sei un traduttore. Se ricevi un testo in italiano, traducilo in inglese. Se ricevi un testo in inglese, traducilo in italiano. Restituisci solo la traduzione, senza commenti."
		},
		{
			"role": "user",
			"content": input_text
		},
	],
)

print(complention.choices[0].message.content)
```
