**Traccia:** Trasforma un set di dati strutturati (come un dizionario Python) in una descrizione testuale completa.

**Obiettivo:** Convertire dati strutturati in una descrizione leggibile.

```python

# TODO Inizializza il client di Open AI

import os, openai
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env') # Read the .env file

# Configure OpenAI client
openai.api_key = os.environ['OPENAI_API_KEY']

client = OpenAI()
```

```python
input_text = input("Enter text: ")
data = {"name": "Giovanni", "age": 30, "city": "Roma", "profession": "Ingegnere"}


complention = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "system",
            "content": "Trasforma il dizionario Python che ricevi in una descrizione testuale fluida e leggibile in italiano."
        },
        {
            "role": "user",
            "content": str(data)
        },
    ],
)

print(complention.choices[0].message.content)
```
